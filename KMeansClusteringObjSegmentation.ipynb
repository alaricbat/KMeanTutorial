```python
import numpy as np
from mnist import MNIST
from sklearn.neighbors import NearestNeighbors
from clazz.CustomKMeans import CustomKMeans
import matplotlib.image as mpimg

import matplotlib.pyplot as plt 
```

### Import IMG

```python
img = mpimg.imread('dataset/Image/girl3.jpg')
plt.imshow(img)
imgplot = plt.imshow(img)
plt.axis('off')
plt.show() 
```

```python
# K-Means requires a 2D input array of shape (Samples, Features).
# This reshapes the 3D color image from (Height, Width, RGB_Channels) 
# into a 2D matrix of shape (Total_Pixels, 3).
# Each pixel becomes a single sample with 3 features (R, G, B color values).
X = img.reshape((img.shape[0]*img.shape[1], img.shape[2]))
print(f'X shape = {X.shape}')
```

```python
for K in [2, 5, 10, 15, 20]:
    kmeans = CustomKMeans(n_clusters=K).fit(X)
    labels = kmeans.predict(X)
    tmp_img = np.zeros_like(X)
    # replace each pixel by its center
    for k in range(K):
        tmp_img[labels == k] = kmeans.centroids[k]
    centers_img = tmp_img.reshape((img.shape[0], img.shape[1], img.shape[2]))
    plt.imshow(centers_img, interpolation='nearest')
    plt.axis('off')
    plt.show()
```
